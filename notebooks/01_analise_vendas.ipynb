# 1. Importação das Bibliotecas

```python
import pandas as pd
```

# 2. Carregamento dos Dados

```python
base = pd.read_csv("../data/raw/train.csv")
```

# 5. Tratamento dos Dados

```python
base["Data do Pedido"] = pd.to_datetime(base["Data do Pedido"], format="%d/%m/%Y")
```

```python
base["Data de Envio"] = pd.to_datetime(base["Data de Envio"], format="%d/%m/%Y")
```

```python
base.info()
```

```python
base = base.dropna(subset=["CEP"])
```

# 6. Análise Exploratória

```python
faturamento_total = base["Vendas"].sum()
print(f"Faturamento total: US$ {faturamento_total:.2f}")
```

```python
quantidade_pedidos = base["ID do Pedido"].nunique()
print(f"Quantidade de pedidos: {quantidade_pedidos}")
```

```python
ticket_medio = faturamento_total / quantidade_pedidos
print(f"Ticket Médio: {ticket_medio:.2f}")
```

```python
vendas_por_categoria = (
    base.groupby("Categoria")["Vendas"].sum().sort_values(ascending=False)
)
print("\nVendas Por Categoria:")
print(vendas_por_categoria)
```

```python
quantidade_por_categoria = base["Categoria"].value_counts()
print("\nQuantidade Vendida por Categoria:")
print(quantidade_por_categoria)
```

```python
vendas_por_segmento = (
    base.groupby("Segmento")["Vendas"].sum().sort_values(ascending=False)
)
print("\nVendas por Segmento:")
print(vendas_por_segmento)
```

```python
vendas_por_região = base.groupby("Região")["Vendas"].sum().sort_values(ascending=False)
print("\nVendas por Região:")
print(vendas_por_região)
```

```python
volume_por_produto = base["Nome do Produto"].value_counts()
print("\nProdutos com maior volume de vendas:")
print(volume_por_produto)
```

```python
faturamento_por_mes = base.groupby(base["Data do Pedido"].dt.to_period("M"))[
    "Vendas"
].sum()
```

```python
periodo_maior_faturamento = faturamento_por_mes.idxmax()
maior_faturamento = faturamento_por_mes.max()
print(f"Maior faturamento: {periodo_maior_faturamento} | US$ {maior_faturamento:.2f}")
```

```python
periodo_menor_faturamento = faturamento_por_mes.idxmin()
menor_faturamento = faturamento_por_mes.min()
print(f"Menor faturamento: {periodo_menor_faturamento} | US$ {menor_faturamento:.2f}")
```

## Principais Tendências e Padrões Identificados

Após a limpeza, o tratamento e a análise exploratória dos dados, foram identificados os seguintes resultados:

### Indicadores Gerais

- **Faturamento total:** US$ 2.252.607,41
- **Quantidade de pedidos:** 4.916
- **Ticket médio:** US$ 458,22

### Desempenho por Categoria

A categoria **Tecnologia** apresentou o maior faturamento, com aproximadamente **US$ 825,9 mil**, seguida por Móveis e Material de Escritório.

Apesar de possuir o menor número de registros de venda entre as categorias, Tecnologia apresentou o maior faturamento. Em contrapartida, **Material de Escritório** possui o maior número de registros, mas o menor faturamento entre as três categorias.

Esse comportamento indica que as vendas da categoria Tecnologia possuem, em média, valores mais elevados.

### Desempenho por Região

A região **Oeste** apresentou o maior valor de vendas, com aproximadamente **US$ 710,2 mil**.

Em seguida aparecem as regiões Leste, Central e Sul. A região **Sul** apresentou o menor faturamento, com aproximadamente **US$ 389,2 mil**.

### Produtos com Maior Volume de Registros

Os produtos que apareceram com maior frequência nos registros de venda foram:

1. Staple envelope — 47 registros
2. Staples — 46 registros
3. Easy-staple paper — 44 registros

Como a base não possui uma coluna de quantidade de unidades, esses valores representam a frequência dos produtos nos registros de venda, e não a quantidade de unidades comercializadas.

### Desempenho ao Longo do Tempo

O período de **maior faturamento** foi **novembro de 2018**, com **US$ 117.938,15**.

O período de **menor faturamento** foi **fevereiro de 2015**, com **US$ 4.519,89**.

### Conclusão

A análise mostrou diferenças importantes entre volume de registros e faturamento. Tecnologia, mesmo aparecendo com menor frequência entre as categorias, apresentou o maior faturamento, enquanto Material de Escritório apresentou comportamento inverso.

Também foi identificada uma diferença relevante no desempenho entre regiões e períodos, que poderá ser explorada visualmente no dashboard desenvolvido no Power BI.
